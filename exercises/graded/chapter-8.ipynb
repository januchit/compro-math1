### 🔧 ระบบตรวจคำตอบ
รันเซลล์ด้านล่าง **หนึ่งครั้ง** ก่อนเริ่มทำ จากนั้นเมื่อทำข้อใดเสร็จ ให้รันเซลล์ `ตรวจ(เลขข้อ)` ที่อยู่ใต้ข้อนั้น เพื่อตรวจว่าคำตอบถูกหรือยัง

```python
# 🔧 ระบบตรวจคำตอบอัตโนมัติ — รันเซลล์นี้ "หนึ่งครั้ง" ก่อนเริ่มทำแบบฝึกหัด
# วิธีใช้: เขียนคำตอบในเซลล์ของแต่ละข้อ กด ▶ รันให้เห็นผลลัพธ์ก่อน
#         จากนั้นรันเซลล์  ตรวจ(เลขข้อ)  ที่อยู่ใต้ข้อนั้น เพื่อตรวจว่าถูกไหม
import io, re, contextlib, json as _json

_KEY = _json.loads(r"""{"1": {"num": [20]}, "2": {"text": ["a", "b", "c"]}, "3": {"num": [1, 2, 3]}, "4": {"num": [240]}, "5": {"num": [10, 20]}, "6": {"num": [100, 2]}, "7": {"text": ["YES"]}, "8": {"num": [3]}, "9": {"num": [1, 2, 3]}, "10": {"num": [7]}, "11": {"num": [1, 2, 3, 4]}, "12": {"num": [1, 3, 4]}, "13": {"num": [1, 2, 3, 4, 5]}, "14": {"num": [3, 4]}, "15": {"num": [1, 3, 5]}, "16": {"num": [2]}, "17": {"num": [0]}, "18": {"text": ["a", "b"], "num": [1, 2]}, "19": {"text": ["YES"]}, "20": {"num": [120]}, "21": {"text": ["'m':2", "'a':2", "'t':2"]}, "22": {"text": ["'the':3"]}, "23": {"text": ["Bob"]}, "24": {"num": [80]}, "25": {"num": [3]}, "26": {"num": [1, 2, 5, 6]}, "27": {"text": ["a", "b", "c"], "num": [80, 90, 70]}, "28": {"num": [5]}, "29": {"text": ["YES"]}, "30": {"num": [3, 4]}, "31": {"num": [3, 4]}, "32": {"num": [1, 3]}, "33": {"text": ["b"]}, "34": {"num": [1, 2, 3, 4]}, "35": {"num": [8]}, "36": {"text": ["YES"]}, "37": {"text": ["1:2", "3:3"]}, "38": {"num": [100]}, "39": {"num": [1, 2, 3, 5]}, "40": {"text": ["a", "c"]}, "41": {"text": ["apple"]}, "42": {"text": ["Eng", "Sci"]}, "43": {"text": ["YES"]}, "44": {"num": [1, 2, 3, 6, 7]}, "45": {"text": ["w"]}, "46": {"num": [2, 4, 6, 8, 1, 3, 5, 7]}, "47": {"text": ["'a':3", "'b':2"]}, "48": {"num": [3, 4]}, "49": {"text": ["b", "c", "a"]}, "50": {"num": [85]}}""")
_KEY = {int(k): v for k, v in _KEY.items()}

def _หาโค้ด(qno):
    marker = "เขียนโค้ดข้อ %d ที่นี่" % qno
    hist = _ih[:-1]                       # ไม่รวมเซลล์ตรวจที่กำลังรันอยู่
    for src in reversed(hist):
        if marker in src:
            return src
    return hist[-1] if hist else None     # สำรอง: เซลล์ที่รันก่อนหน้า

def _ไม่มีโค้ด(code):
    if not code:
        return True
    body = "\n".join(l for l in code.splitlines() if not l.strip().startswith("#"))
    return body.strip() == ""

def _ตัวเลข(s):
    return [float(x) for x in re.findall(r"-?\d+\.?\d*", s)]

def _นอร์ม(s):
    return s.replace(" ", "").replace("\t", "").replace("\n", "").lower()

def ตรวจ(qno):
    item = _KEY.get(qno)
    if item is None:
        print("ยังไม่มีเฉลยของข้อ", qno); return
    code = _หาโค้ด(qno)
    if _ไม่มีโค้ด(code):
        print("⚠️  ข้อ %d: ยังไม่พบโค้ดคำตอบ — กรุณารันเซลล์คำตอบของข้อนี้ก่อน แล้วค่อยรันเซลล์ตรวจอีกครั้ง" % qno)
        return
    try:
        buf = io.StringIO()
        with contextlib.redirect_stdout(buf):
            exec(compile(code, "<คำตอบข้อ %d>" % qno, "exec"), {})
        out = buf.getvalue()
    except Exception as e:
        print("❌  ข้อ %d: โค้ดรันแล้วเกิดข้อผิดพลาด → %s: %s" % (qno, type(e).__name__, e))
        return

    if item.get("manual"):
        if out.strip():
            print("ℹ️  ข้อ %d: ข้อนี้ขึ้นกับข้อมูลของนักศึกษา ระบบตรวจอัตโนมัติให้ไม่ได้" % qno)
            print("    โปรแกรมของคุณแสดงผลว่า:")
            for line in out.rstrip().splitlines():
                print("      ", line)
            print("    ✔️  ถ้าผลตรงกับที่โจทย์ต้องการ ถือว่าผ่าน")
        else:
            print("⚠️  ข้อ %d: ยังไม่มีผลลัพธ์แสดงออกมา ลองตรวจว่าใช้ print() แล้วหรือยัง" % qno)
        return

    problems = []
    if "text" in item:
        n = _นอร์ม(out)
        for t in item["text"]:
            if _นอร์ม(t) not in n:
                problems.append("ต้องมีข้อความ: %s" % t)
    if "num" in item:
        nums = _ตัวเลข(out)
        tol = item.get("tol")
        for v in item["num"]:
            ok = any(abs(v - x) <= (tol if tol is not None else max(1e-6, abs(v) * 1e-4)) for x in nums)
            if not ok:
                problems.append("ต้องมีค่า: %s" % (("%g" % v)))

    if not problems:
        print("✅  ข้อ %d: ถูกต้อง! เก่งมาก 🎉" % qno)
    else:
        print("❌  ข้อ %d: ยังไม่ถูก" % qno)
        for p in problems:
            print("     • " + p)
        shown = out.strip() if out.strip() else "(ไม่มีผลลัพธ์ — ใช้ print() หรือยัง?)"
        print("     ผลลัพธ์ที่โปรแกรมของคุณแสดง: %s" % shown)

print("✅ พร้อมตรวจคำตอบแล้ว — ทำข้อไหนเสร็จ ให้รันเซลล์  ตรวจ(เลขข้อ)  ใต้ข้อนั้นได้เลย")
```
